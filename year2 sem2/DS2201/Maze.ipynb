```python
import random

class MazeGenerator:
    def __init__(self, rows, cols):
        self.rows = rows
        self.cols = cols
        self.maze = [[{'N': True, 'S': True, 'E': True, 'W': True} 
                     for _ in range(cols)] for _ in range(rows)]
        
        # 2-d arrays 

        self.visited = [[False for _ in range(cols)] for _ in range(rows)]
        # visited cell is also a 2-d array
        
    def generate(self, start_row=0, start_col=0):
        """使用栈（DFS）生成迷宫"""
        stack = [(start_row, start_col)]
        self.visited[start_row][start_col] = True
        
        while stack:
            current_row, current_col = stack.pop()
            neighbors = self.get_unvisited_neighbors(current_row, current_col)
            # Check if neighbor exist
            
            if neighbors:
                # 将当前单元格重新压入栈中，以便回溯
                stack.append((current_row, current_col))
                
                # 随机选择一个未访问的邻居
                next_row, next_col, direction = random.choice(neighbors)
                # a neighbor will pass three dimension out
                
                # 移除墙壁
                self.remove_wall(current_row, current_col, next_row, next_col, direction)
                
                # 标记邻居为已访问并压入栈
                self.visited[next_row][next_col] = True
                stack.append((next_row, next_col))
    
    def get_unvisited_neighbors(self, row, col):
        """获取未访问的邻居单元格"""
        neighbors = []
        directions = [
            ('N', -1, 0),  # 北
            ('S', 1, 0),   # 南
            ('W', 0, -1),  # 西
            ('E', 0, 1)    # 东
        ]
        
        for direction, dr, dc in directions:
            new_row, new_col = row + dr, col + dc
            
            if (0 <= new_row < self.rows and 
                0 <= new_col < self.cols and 
                not self.visited[new_row][new_col]):
                neighbors.append((new_row, new_col, direction))
        
        return neighbors
    
    def remove_wall(self, r1, c1, r2, c2, direction):
        """移除两个单元格之间的墙壁"""
        # 移除当前单元格的墙壁
        if direction == 'N':
            self.maze[r1][c1]['N'] = False
            self.maze[r2][c2]['S'] = False
        elif direction == 'S':
            self.maze[r1][c1]['S'] = False
            self.maze[r2][c2]['N'] = False
        elif direction == 'W':
            self.maze[r1][c1]['W'] = False
            self.maze[r2][c2]['E'] = False
        elif direction == 'E':
            self.maze[r1][c1]['E'] = False
            self.maze[r2][c2]['W'] = False
    
    def print_maze(self):
        """打印迷宫（文本表示）"""
        # 打印顶部边界
        print("+" + "---+" * self.cols)
        
        for row in range(self.rows):
            # 打印西边界和单元格内容
            print("|", end="")
            for col in range(self.cols):
                # 打印东墙壁
                if self.maze[row][col]['E']:
                    print("   |", end="")
                else:
                    print("    ", end="")
            print()
            
            # 打印南墙壁
            print("+", end="")
            for col in range(self.cols):
                if self.maze[row][col]['S']:
                    print("---+", end="")
                else:
                    print("   +", end="")
            print()

# 使用示例
if __name__ == "__main__":
    maze = MazeGenerator(5, 5)
    maze.generate()
    print("生成的迷宫：")
    maze.print_maze()
```

```python
import random

class Maze:
    """完整的迷宫类，包含生成和解决功能"""
    
    def __init__(self, rows, cols):
        """初始化迷宫"""
        self.rows = rows
        self.cols = cols
        # 初始化迷宫结构
        self.maze = [[{'N': True, 'S': True, 'E': True, 'W': True} 
                     for _ in range(cols)] for _ in range(rows)]
        self.visited = [[False for _ in range(cols)] for _ in range(rows)]
        self.path = []  # 存储找到的路径
        self.solution_visited = None  # 解决时的访问记录
        
    def generate(self, start_row=0, start_col=0):
        """生成迷宫（使用栈的DFS算法）"""
        print(f"开始生成 {self.rows}×{self.cols} 迷宫...")
        
        stack = [(start_row, start_col)]
        self.visited[start_row][start_col] = True
        
        total_cells = self.rows * self.cols
        visited_count = 1
        
        while stack:
            current_row, current_col = stack.pop()
            neighbors = self._get_unvisited_neighbors(current_row, current_col)
            
            if neighbors:
                stack.append((current_row, current_col))
                next_row, next_col, direction = random.choice(neighbors)
                
                # 移除墙壁
                self._remove_wall(current_row, current_col, next_row, next_col, direction)
                
                self.visited[next_row][next_col] = True
                visited_count += 1
                stack.append((next_row, next_col))
                
                # 显示进度
                if visited_count % max(1, total_cells // 10) == 0:
                    progress = visited_count / total_cells * 100
                    print(f"  生成进度: {progress:.0f}%")
        
        print("迷宫生成完成！")
        return self.maze
    
    def _get_unvisited_neighbors(self, row, col):
        """获取未访问的邻居"""
        neighbors = []
        directions = [
            ('N', -1, 0),
            ('S', 1, 0),
            ('W', 0, -1),
            ('E', 0, 1)
        ]
        
        for direction, dr, dc in directions:
            new_row, new_col = row + dr, col + dc
            if (0 <= new_row < self.rows and 
                0 <= new_col < self.cols and 
                not self.visited[new_row][new_col]):
                neighbors.append((new_row, new_col, direction))
        
        return neighbors
    
    def _remove_wall(self, r1, c1, r2, c2, direction):
        """移除两个单元格之间的墙壁"""
        if direction == 'N':
            self.maze[r1][c1]['N'] = False
            self.maze[r2][c2]['S'] = False
        elif direction == 'S':
            self.maze[r1][c1]['S'] = False
            self.maze[r2][c2]['N'] = False
        elif direction == 'W':
            self.maze[r1][c1]['W'] = False
            self.maze[r2][c2]['E'] = False
        elif direction == 'E':
            self.maze[r1][c1]['E'] = False
            self.maze[r2][c2]['W'] = False
    
    def solve(self, start_row=0, start_col=0, end_row=None, end_col=None):
        """解决迷宫（使用栈的DFS算法）"""
        if end_row is None:
            end_row = self.rows - 1
        if end_col is None:
            end_col = self.cols - 1
            
        print(f"\n开始解决迷宫...")
        print(f"起点: ({start_row}, {start_col}), 终点: ({end_row}, {end_col})")
        
        # 初始化解决时的访问记录
        self.solution_visited = [[False for _ in range(self.cols)] for _ in range(self.rows)]
        
        # 使用栈进行DFS搜索
        stack = [(start_row, start_col, [])]  # (row, col, path)
        
        step_count = 0
        while stack:
            row, col, current_path = stack.pop()
            new_path = current_path + [(row, col)]
            step_count += 1
            
            # 检查是否到达终点
            if row == end_row and col == end_col:
                self.path = new_path
                print(f"找到路径！总步数: {step_count}, 路径长度: {len(self.path)}")
                return True
            
            # 如果已访问过，跳过
            if self.solution_visited[row][col]:
                continue
                
            self.solution_visited[row][col] = True
            
            # 获取可以走的方向
            possible_directions = self._get_possible_directions(row, col)
            
            # 将每个可能的方向加入栈
            for next_row, next_col, _ in possible_directions:
                if not self.solution_visited[next_row][next_col]:
                    stack.append((next_row, next_col, new_path))
        
        print("没有找到路径！")
        return False
    
    def _get_possible_directions(self, row, col):
        """获取当前位置可以走的方向（没有墙壁的方向）"""
        possible = []
        
        # 检查北边
        if row > 0 and not self.maze[row][col]['N']:
            possible.append((row-1, col, 'N'))
        
        # 检查南边
        if row < self.rows-1 and not self.maze[row][col]['S']:
            possible.append((row+1, col, 'S'))
        
        # 检查西边
        if col > 0 and not self.maze[row][col]['W']:
            possible.append((row, col-1, 'W'))
        
        # 检查东边
        if col < self.cols-1 and not self.maze[row][col]['E']:
            possible.append((row, col+1, 'E'))
            
        return possible
    
    def print_maze_with_path(self):
        """打印迷宫和找到的路径"""
        if not self.path:
            print("还没有找到路径，请先调用 solve() 方法")
            return
            
        print("\n" + "=" * 60)
        print(f"迷宫 ({self.rows}×{self.cols}) 和找到的路径")
        print("=" * 60)
        
        # 创建一个标记数组，标记路径上的点
        path_set = set(self.path)
        
        # 起点和终点
        start = self.path[0]
        end = self.path[-1]
        
        # 打印迷宫顶部边界
        print("+" + "---+" * self.cols)
        
        for row in range(self.rows):
            # 打印西边界和单元格内容
            print("|", end="")
            for col in range(self.cols):
                # 检查是否是路径上的点
                if (row, col) == start:
                    symbol = "S"  # 起点
                elif (row, col) == end:
                    symbol = "E"  # 终点
                elif (row, col) in path_set:
                    symbol = "·"  # 路径点
                else:
                    symbol = " "  # 空单元格
                
                print(f" {symbol} ", end="")
                
                # 打印东墙壁
                if self.maze[row][col]['E']:
                    print("|", end="")
                else:
                    print(" ", end="")
            
            print()
            
            # 打印南墙壁（最后一行特殊处理）
            print("+", end="")
            for col in range(self.cols):
                if row == self.rows - 1:
                    # 最后一行打印完整的南墙壁
                    if self.maze[row][col]['S']:
                        print("---+", end="")
                    else:
                        print("   +", end="")
                else:
                    # 其他行打印下一行的北墙壁
                    if self.maze[row+1][col]['N']:
                        print("---+", end="")
                    else:
                        print("   +", end="")
            print()
    
    def print_detailed_path(self):
        """打印详细的路径信息"""
        if not self.path:
            print("还没有找到路径")
            return
            
        print("\n" + "=" * 60)
        print("详细路径信息")
        print("=" * 60)
        print(f"路径总长度: {len(self.path)} 步")
        print(f"起点: {self.path[0]}")
        print(f"终点: {self.path[-1]}")
        
        print("\n路径坐标:")
        for i, (row, col) in enumerate(self.path):
            # 计算方向（如果有上一个点）
            direction = ""
            if i > 0:
                prev_row, prev_col = self.path[i-1]
                if row < prev_row:
                    direction = "↑ 北"
                elif row > prev_row:
                    direction = "↓ 南"
                elif col < prev_col:
                    direction = "← 西"
                elif col > prev_col:
                    direction = "→ 东"
            
            if i == 0:
                print(f"第{i+1:2d}步: ({row}, {col}) [起点]")
            elif i == len(self.path) - 1:
                print(f"第{i+1:2d}步: ({row}, {col}) {direction} [终点]")
            else:
                print(f"第{i+1:2d}步: ({row}, {col}) {direction}")
    
    def print_statistics(self):
        """打印迷宫统计信息"""
        print("\n" + "=" * 60)
        print("迷宫统计信息")
        print("=" * 60)
        
        # 计算墙壁数量
        total_walls = 0
        open_walls = 0
        
        for row in range(self.rows):
            for col in range(self.cols):
                for direction in ['N', 'S', 'E', 'W']:
                    total_walls += 1
                    if not self.maze[row][col][direction]:
                        open_walls += 1
        
        # 每个单元格有4面墙，但每面墙被两个单元格共享
        actual_walls = total_walls / 2
        actual_open_walls = open_walls / 2
        wall_density = (actual_walls - actual_open_walls) / actual_walls * 100
        
        print(f"迷宫大小: {self.rows} × {self.cols}")
        print(f"总单元格数: {self.rows * self.cols}")
        print(f"墙壁总数: {int(actual_walls)}")
        print(f"打开的门数: {int(actual_open_walls)}")
        print(f"墙壁密度: {wall_density:.1f}%")
        
        if self.path:
            print(f"\n解决信息:")
            print(f"  找到路径: 是")
            print(f"  路径长度: {len(self.path)} 步")
            print(f"  起点到终点曼哈顿距离: {abs(self.path[0][0]-self.path[-1][0]) + abs(self.path[0][1]-self.path[-1][1])}")
            
            # 计算解决效率
            visited_count = 0
            if self.solution_visited:
                visited_count = sum(sum(row) for row in self.solution_visited)
            efficiency = len(self.path) / visited_count * 100 if visited_count > 0 else 0
            print(f"  访问的单元格数: {visited_count}")
            print(f"  解决效率: {efficiency:.1f}% (路径长度/访问单元格数)")
        else:
            print(f"\n解决信息: 未解决")


def create_custom_maze():
    """创建自定义大小的迷宫并解决"""
    print("迷宫生成和解决程序")
    print("=" * 60)
    
    # 获取用户输入
    while True:
        try:
            rows = int(input("请输入迷宫的行数 (推荐5-20): "))
            cols = int(input("请输入迷宫的列数 (推荐5-20): "))
            if rows > 0 and cols > 0:
                break
            else:
                print("行数和列数必须为正整数！")
        except ValueError:
            print("请输入有效的数字！")
    
    # 创建迷宫
    maze = Maze(rows, cols)
    
    # 生成迷宫
    maze.generate()
    
    # 解决迷宫
    maze.solve()  # 默认从(0,0)到(rows-1, cols-1)
    
    # 显示结果
    maze.print_maze_with_path()
    maze.print_detailed_path()
    maze.print_statistics()
    
    return maze


def run_demo():
    """运行演示程序"""
    print("迷宫生成和解决演示程序")
    print("=" * 60)
    
    # 创建不同大小的迷宫进行演示
    demo_sizes = [(5, 5), (8, 8), (10, 10)]
    
    for i, (rows, cols) in enumerate(demo_sizes, 1):
        print(f"\n演示 {i}: {rows}×{cols} 迷宫")
        print("-" * 40)
        
        maze = Maze(rows, cols)
        maze.generate()
        
        # 尝试不同的起点和终点
        if maze.solve(0, 0, rows-1, cols-1):
            maze.print_maze_with_path()
            print(f"路径长度: {len(maze.path)}")
        else:
            print("无法找到路径！")
        
        print()


def interactive_mode():
    """交互式模式"""
    print("交互式迷宫生成和解决")
    print("=" * 60)
    
    while True:
        print("\n选项:")
        print("1. 创建新迷宫并解决")
        print("2. 运行演示")
        print("3. 退出")
        
        choice = input("请选择 (1-3): ").strip()
        
        if choice == '1':
            maze = create_custom_maze()
            
            # 提供额外选项
            while True:
                print("\n额外选项:")
                print("1. 尝试不同的起点和终点")
                print("2. 重新显示迷宫和路径")
                print("3. 返回主菜单")
                
                sub_choice = input("请选择 (1-3): ").strip()
                
                if sub_choice == '1':
                    try:
                        start_row = int(input("请输入起点行号 (0-{}): ".format(maze.rows-1)))
                        start_col = int(input("请输入起点列号 (0-{}): ".format(maze.cols-1)))
                        end_row = int(input("请输入终点行号 (0-{}): ".format(maze.rows-1)))
                        end_col = int(input("请输入终点列号 (0-{}): ".format(maze.cols-1)))
                        
                        if (0 <= start_row < maze.rows and 0 <= start_col < maze.cols and
                            0 <= end_row < maze.rows and 0 <= end_col < maze.cols):
                            maze.path = []  # 清除旧路径
                            maze.solution_visited = None
                            if maze.solve(start_row, start_col, end_row, end_col):
                                maze.print_maze_with_path()
                                maze.print_detailed_path()
                            else:
                                print("无法找到从起点到终点的路径！")
                        else:
                            print("坐标超出范围！")
                    except ValueError:
                        print("请输入有效的数字！")
                        
                elif sub_choice == '2':
                    maze.print_maze_with_path()
                    maze.print_detailed_path()
                    
                elif sub_choice == '3':
                    break
                    
                else:
                    print("无效选择！")
                    
        elif choice == '2':
            run_demo()
            
        elif choice == '3':
            print("感谢使用，再见！")
            break
            
        else:
            print("无效选择，请重新选择！")


if __name__ == "__main__":
    # 直接运行演示
    print("迷宫生成和解决程序")
    print("=" * 60)
    
    # 创建一个中等大小的迷宫进行演示
    print("正在生成 10×10 迷宫...")
    
    maze = Maze(10, 10)
    maze.generate()
    
    print("\n正在解决迷宫...")
    if maze.solve(0, 0, 9, 9):  # 从左上到右下
        print("\n迷宫和路径:")
        maze.print_maze_with_path()
        
        print("\n路径详细信息:")
        maze.print_detailed_path()
        
        print("\n统计信息:")
        maze.print_statistics()
        
        print("\n" + "=" * 60)
        print("提示: 运行 interactive_mode() 进入交互式模式")
        print("或者运行 create_custom_maze() 创建自定义迷宫")
        print("=" * 60)
    else:
        print("无法找到路径！")
```
