```python
# Assignment3 HU Yuanzhe 24107147d
# Question1 C)
def calculate_asymmetric_cost(villages, station_x):
    """
    Calculates total cost for a specific station location.
    
    Args:
        villages: List of village locations. E.g. [2, 5, 12, 18, 20].
        station_x: Location of the power station.
    
    Returns:
        Total installation cost (float).
    """
    total = 0.0
    
    for village in villages:
        if village < station_x:
            # West of station: 3 per km
            total += 3 * (station_x - village)
        elif village > station_x:
            # East of station: 1 per km
            total += 1 * (village - station_x)
        # village == station_x: cost is 0
    
    return total
```

```python
# Question1 D)
def solve_greedy(villages):
    """
    Finds the optimal station location.
    
    Args:
        villages: List of village locations. E.g. [2, 5, 12, 18, 20].
    
    Returns:
        Tuple containing (optimal_station, min_cost).
    """
    # Sort for consistent order
    villages_sorted = sorted(villages)
    
    best_loc = None
    best_cost = float('inf')
    
    for station in villages_sorted:
        cost = calculate_asymmetric_cost(villages_sorted, station)
        if cost < best_cost:
            best_cost = cost
            best_loc = station
    
    return best_loc, best_cost


# Test
if __name__ == "__main__":
    test_data = [2, 5, 12, 18, 20]
    station, cost = solve_greedy(test_data)
    print(f"Optimal station: {station}")
    print(f"Minimum cost: {cost}")
```

```python
# Question2 def cost(p, i, villages):
def cost(p, i, villages):
    """
    Calculate the minimum cost to serve villages from index p to i (inclusive) with a single station.
    
    Args:
        p: Starting index (inclusive) in villages list.
        i: Ending index (inclusive) in villages list.
        villages: Sorted list of village coordinates.
    
    Returns:
        Minimum cost for this segment.
    """
    # Get the segment of villages
    segment = villages[p:i+1]
    
    # Find optimal station for this segment and get its cost
    _, segment_cost = solve_greedy(segment)
    
    return segment_cost
```
