```python
# quadratic time
def minMax_quadratic(numbers):
    min_number = None
    max_number = None
    for i in range(len(numbers)):
        isMin = True
        isMax = True
        for j in range(len(numbers)):
            if j != i:
                # 只要发现有别的数比它大，它就不是最大值
                if numbers[j] > numbers[i]:
                    isMax = False
                # 只要发现有别的数比它小，它就不是最小值
                if numbers[j] < numbers[i]:
                    isMin = False

                # 经过一整轮比对，如果依然判定为True，则赋值
                if isMin:
                    min_number = numbers[i]
                if isMax:
                    max_number = numbers[i]
                return (min_number, max_number)
```

```python
class Queue:
    def __init__(self, size): # constructor
        self.items = [None] * size
        self.front = 0 # index of the front slot 
        self.rear = 0 # index of the first empty slot at the rear of the queue 

    def enqueue(self, item): # add an item to the rear of the queue
        if (self.rear +1) % len(self.items) == self.front:
            raise Exception("Queue id full")
        self.items[self.rear] = item 
        self.rear = (self.rear +1) % len(self.items)

    def dequeue(self): # removes and returns the first item of the queue
        if self.front == self.rear:
            raise Exception("Queue is empty")
        popped = self.items[self.front]
        self.front = (self.front + 1) % len(self.items)
        return popped

    def front(self): # return the front element 
        if self.front != self.rear:
            return self.items[self.front]

    def isEmpty(self): # 当头尾指针相遇时 队列为空
        return self.front == self.rear

    def __len__(self): # 利用取模运算计算循环数组中的两个指针的实际距离
        return (len(self.items) - self.front + self.rear) % len(self.items)
```

```python
def magic_movement(k, numbers):
    n = len(numbers)
    q = Queue(n + 1)  # 1. 创建队列，容量比数据多1（防止满队判断出错）

    # 2. 初始化队列：把每个数字和它的原始位置索引一起放进队列
    for i in range(n):
        q.enqueue((numbers[i], i))  # 重要：这里要用 enqueue，不是 queue

    result_items = []  # 3. 用来保存结果的列表（你之前写的是 results_items，后面用 result_items，要统一）

    while not q.isEmpty():  # 4. 只要队列不空就继续处理
        removed_items = []  # 5. 临时列表，存这一轮取出的k个元素

        # 6. 取 min(k, 队列当前长度) 个元素，避免取多了
        for i in range(min(k, len(q))):
            removed_items.append(q.dequeue())  # 从队列头部取出

        # 7. 如果取出的数量小于k（说明是最后一批了）
        if len(removed_items) < k:
            for item in removed_items:
                result_items.append(item[1])  # 8. 直接记录它们的原始索引
        else:
            # 9. 找这批里面数字最大的那个
            largest_index = 0
            for i in range(1, len(removed_items)):
                # 10. 比较数字部分：removed_items[i][0] 是数字
                if removed_items[i][0] > removed_items[largest_index][0]:
                    largest_index = i

            # 11. 把最大数字的原始索引加入结果
            result_items.append(removed_items[largest_index][1])

            # 12. 删除这个最大的元素（因为它已经被处理了）
            del removed_items[largest_index]

            # 13. 剩下的元素：数字减1，然后重新放回队列尾部
            for item in removed_items:
                if item[0] > 1:  # 14. 只有数字大于1才需要放回（减到0就不放了）
                    q.enqueue((item[0] - 1, item[1]))

    return result_items  # 15. 返回所有被处理的元素的原始索引顺序

# 测试
print(magic_movement(2, [17, 18, 20]))  # 你的第一个例子 [17-20] 其实是三个数：17, 18, 20
print(magic_movement(3, [17, 18, 20, 23, 24]))
```

```python
class NaiveArrayQueue:
    def __init__(self, capacity):
        self.queue = [None] * capacity
        self.capacity = capacity
        self.rear = 0  # 下一个插入位置
        
    def enqueue(self, item):
        if self.rear == self.capacity:
            raise Exception("Queue is full")
        self.queue[self.rear] = item
        self.rear += 1
        
    def dequeue(self):
        if self.rear == 0:
            raise Exception("Queue is empty")
        # ❌ 问题所在：所有元素前移，O(n)时间
        item = self.queue[0]
        for i in range(1, self.rear):
            self.queue[i-1] = self.queue[i]
        self.rear -= 1
        return item
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedListQueue:
    def __init__(self):
        self.head = None  # 队首
        self.tail = None  # 队尾
        
    def enqueue(self, item):
        new_node = Node(item)
        if self.tail:  # 如果队列不空
            self.tail.next = new_node
        else:  # 如果队列空
            self.head = new_node
        self.tail = new_node
        
    def dequeue(self):
        if not self.head:
            raise Exception("Queue is empty")
        item = self.head.data
        self.head = self.head.next
        if not self.head:  # 如果队列变空
            self.tail = None
        return item
```

```python
class CircularArrayQueue:
    def __init__(self, capacity):
        self.queue = [None] * capacity
        self.capacity = capacity
        self.front = 0  # 队首位置
        self.rear = 0   # 队尾位置（下一个插入位置）
        self.size = 0   # 当前元素个数
        
    def enqueue(self, item):
        if self.is_full():
            raise Exception("Queue is full")
        self.queue[self.rear] = item
        self.rear = (self.rear + 1) % self.capacity  # 循环的关键
        self.size += 1
        
    def dequeue(self):
        if self.is_empty():
            raise Exception("Queue is empty")
        item = self.queue[self.front]
        self.front = (self.front + 1) % self.capacity  # 循环的关键
        self.size -= 1
        return item
        
    def is_empty(self):
        return self.size == 0
        
    def is_full(self):
        return self.size == self.capacity
```

```python
# 爬楼梯问题
def count_ways(n):
    if n == 1: # 只有1阶 只有一种方法
        return 1
    if n == 2: # 有两阶 1步+1步 或者直接2步
        return 2
    return count_ways(n-1) + count_ways(n-2) # 递归步骤
```

```python
# 生成字符串所有排列：
def generatePermutations(s): # 基线条件：字符串条件<=1 其排列就是本身
    if len(s) <= 1:
        return [s] if s else [] # 这里文档代码返回s，但更严谨应该返回包含s的列表
        perms = []
        for i in range(len(s)): # 1 选择一个字符作为当前排列的第一个字符
            char = s[i]
            remaining = s[:i] + s[i+1:]#2 获取剩下的字符
            for p in generatePermutations(remaining):# 3 递归生成剩下字符的所有排列
                perms.append(char + p)
            return perms
```

```python
#DFS Pre-order
def depthFirstSearch(root, target_value):
    if not root:
        return False
    stack = [root] # 使用列表模拟栈
    while stack:
        node = stack.pop()
        if node.data == target_value: # 找到目标
            return True #先将右子节点入栈，再将左子节点入栈，保证左子节点先出栈(Pre-order)
        if node.right:
            stack.append(node.right)
        if node.left:
            stack.append(node.left)
        return False
```

```python
# BFS Queue
def breathFirstSearch(root, target_value):
    if not root:
        return False
    from collection import deque
    queue = deque([root])
    while queue:
        node = queue.popleft()
        if node.data == target_value: # 找到目标
            return True
        if node.left:
            queue.append(node.left)
        if node.right:
            queue.append(node.right)
        return False
```

```python
# BST Core steps
# 1. Find/Search
def findBST(node, key):
    if node is None: # 1.基准情况1:未找到
        return None
    if node.key == key: # 2.基准情况2:找到
        return node.value
    #递归步骤
    if key < node.key:
        return findBST(node.left, key) # 在左子树中查找
    else: # key>node.key
        return findBST(node.right, key) # 在右子树中查找
#2.Insert
def insertInBST(node, key, value):
    if node is None: # 基准情况：找到插入位置
        return BSTNode(key, value)
    #递归步骤
    if key < node.key:
        node.left = insertInBST(node.left, key, value)
    elif key > node.key: # 通常处理key不相等的情况
        node.right = insertInBST(node.right, key, value)
    #如果key相等，可以根据需求处理（如更新value）
    return node
# 3.Remove
def removalInBST(node, key):
    if node is None:
        return None
    if key < node.key:
        node.left = removalInBST(node.left, key)
    elif key > node.key:
        node.right = removalInBST(node.right, key)
    else: # 找到要删除的节点
        # 情况1&2:0个或1个节点
        if node.left is None:
            return node.right
        elif node.right is None:
            return node.left
        # 情况3:2个子节点
        else:
            #1.找到中序后继（右子树最小节点）
            successor = findMin(node.right)
            #2.用后继的值替换当前节点
            node.key, node.value = successor.key, successor.value
            #3.删除右子树中的那个后继节点
            node.right = removalInBST(node.right, succeesor.key)
        return node

def findMin(node):
    current = node
    while current.left is not None:
        current = current.left
    return current
```
