```python
# guess what this game is about
from random import randint
mychoice = randint(1,3)
if mychoice == 1:
    me = "p"
if mychoice == 2:
    me = "s"
if mychoice == 3:
    me = "r"
# you need to enter p or s or r
you = input("Please enter your choice (p / s / r): ")
# what would happen if you enter something else?
print("You choose", you)
print("I choose", me)
if you == me:
    print("This is a draw")
if you == "p" and me == "s":
    print("I win")
if you == "p" and me == "r":
    print("You win")
if you == "s" and me == "p":
    print("You win")
if you == "s" and me == "r":
    print("I win")
if you == "r" and me == "p":
    print("I win")
if you == "r" and me == "s":
    print("You win")
```

```python
#Q7
import math
# Given sides of the triangle
a = 15
b = 24
c = 20
# Method A: Compute Area First, Then Altitudes, Then Angles
s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))
h_a = (2 * area) / a
h_b = (2 * area) / b
h_c = (2 * area) / c
A = math.degrees(math.acos((b**2 + c**2 - a**2) / (2 * b * c)))
B = math.degrees(math.acos((a**2 + c**2 - b**2) / (2 * a * c)))
C = math.degrees(math.acos((a**2 + b**2 - c**2) / (2 * a * b)))
print(f"Method A:")
print(f"Area of the triangle: {area:.2f}")
print(f"Altitude to side a: {h_a:.2f}")
print(f"Altitude to side b: {h_b:.2f}")
print(f"Altitude to side c: {h_c:.2f}")
print(f"Angle A: {A:.2f} degrees")
print(f"Angle B: {B:.2f} degrees")
print(f"Angle C: {C:.2f} degrees")
print()
```
